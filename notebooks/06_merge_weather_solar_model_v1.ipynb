```python
from pathlib import Path
import pandas as pd

def find_project_root(start_path):
    current = Path(start_path).resolve()
    
    for parent in [current] + list(current.parents):
        if (parent / "data").exists() and (parent / "notebooks").exists():
            return parent
    
    raise FileNotFoundError("Impossible de trouver la racine du projet EnerAI-Africa.")

PROJECT_ROOT = find_project_root(Path.cwd())

DATA_RAW = PROJECT_ROOT / "data" / "raw"
DATA_PROCESSED = PROJECT_ROOT / "data" / "processed"

print("Dossier courant Jupyter :", Path.cwd().resolve())
print("Racine du projet        :", PROJECT_ROOT)
print("DATA_RAW                :", DATA_RAW)

consumption_path = DATA_RAW / "consumption" / "household_data_60min_singleindex.csv"
meteo_path = DATA_RAW / "meteo" / "ndjamena_meteo.csv"
solar_path = DATA_RAW / "solar" / "POWER_Point_Hourly_20010401_20260601_012d11N_015d05E_LST.csv"

print("Consommation :", consumption_path.exists(), consumption_path)
print("Météo        :", meteo_path.exists(), meteo_path)
print("Solaire      :", solar_path.exists(), solar_path)

if not consumption_path.exists():
    raise FileNotFoundError(f"Fichier consommation introuvable : {consumption_path}")

if not meteo_path.exists():
    raise FileNotFoundError(f"Fichier météo introuvable : {meteo_path}")

if not solar_path.exists():
    raise FileNotFoundError(f"Fichier solaire introuvable : {solar_path}")

df_conso = pd.read_csv(consumption_path)
df_meteo = pd.read_csv(meteo_path)
df_solar = pd.read_csv(solar_path)

print("Chargement réussi")
print("df_conso :", df_conso.shape)
print("df_meteo :", df_meteo.shape)
print("df_solar :", df_solar.shape)
```
